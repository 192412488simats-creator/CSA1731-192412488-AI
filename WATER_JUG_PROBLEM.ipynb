```python

# Water Jug Problem using BFS

from collections import deque

def water_jug(cap1, cap2, target):
    queue = deque([(0, 0, [])])
    visited = set()

    while queue:
        a, b, path = queue.popleft()

        if (a, b) in visited:
            continue

        visited.add((a, b))
        path = path + [(a, b)]

        if a == target or b == target:
            return path

        pour1 = min(a, cap2 - b)
        pour2 = min(b, cap1 - a)

        states = [
            (cap1, b),             # Fill jug 1
            (a, cap2),             # Fill jug 2
            (0, b),                # Empty jug 1
            (a, 0),                # Empty jug 2
            (a - pour1, b + pour1),# Pour jug 1 to jug 2
            (a + pour2, b - pour2) # Pour jug 2 to jug 1
        ]

        for state in states:
            if state not in visited:
                queue.append((state[0], state[1], path))

    return None

result = water_jug(4, 3, 2)

if result:
    print("Solution:")
    for state in result:
        print(state)
else:
    print("No solution found")
```
