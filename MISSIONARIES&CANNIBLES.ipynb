```python

# Missionaries and Cannibals using BFS

from collections import deque

def valid(m, c):
    if m < 0 or c < 0 or m > 3 or c > 3:
        return False

    if m > 0 and m < c:
        return False

    mr = 3 - m
    cr = 3 - c

    if mr > 0 and mr < cr:
        return False

    return True

def solve():
    start = (3, 3, 0)
    goal = (0, 0, 1)

    moves = [(1, 0), (2, 0), (0, 1),
             (0, 2), (1, 1)]

    queue = deque([(start, [])])
    visited = {start}

    while queue:
        state, path = queue.popleft()

        if state == goal:
            return path + [state]

        m, c, boat = state

        for x, y in moves:
            if boat == 0:
                new_state = (m-x, c-y, 1)
            else:
                new_state = (m+x, c+y, 0)

            if valid(new_state[0], new_state[1]):
                if new_state not in visited:
                    visited.add(new_state)
                    queue.append(
                        (new_state, path + [state])
                    )

    return None

solution = solve()

if solution:
    print("Solution:")
    for state in solution:
        print(state)
else:
    print("No solution found")
```
